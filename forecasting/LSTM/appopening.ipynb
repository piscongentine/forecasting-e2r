```python
import pandas as pd
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
import matplotlib.pyplot as plt
from io import StringIO # Keep StringIO for embedding data directly

# --- Simplified Workflow: NLP & Forecasting ---
# This workflow outlines essential steps for building predictive models.
# It's applicable for both text-based (NLP) and time-based (Forecasting) data.

# --- 1. Prepare Your Data ---
print("--- 1. Prepare Your Data ---")
df = pd.read_csv(StringIO(csv_data))

df['whatsapp_opened'] = df['whatsapp_opened'].fillna(0).astype(int)
df['snapchat_opened'] = df['snapchat_opened'].fillna(0).astype(int)
df['timestamp'] = pd.to_datetime(df['timestamp'])
df.set_index('timestamp', inplace=True)
df['total_app_opened'] = df['whatsapp_opened'] + df['snapchat_opened']
daily_data = df['total_app_opened'].resample('D').sum().reset_index()
daily_data.rename(columns={'timestamp': 'Date', 'total_app_opened': 'Value'}, inplace=True)

n_lags = 3 # Use last 3 days for prediction

def create_sequences(data, n_lags):
    X, y = [], []
    for i in range(len(data) - n_lags):
        X.append(data[i:(i + n_lags)])
        y.append(data[i + n_lags])
    return np.array(X), np.array(y)

scaler = MinMaxScaler(feature_range=(0, 1))
scaled_data = scaler.fit_transform(daily_data['Value'].values.reshape(-1, 1))

X, y = create_sequences(scaled_data, n_lags)
X = X.reshape(X.shape[0], X.shape[1], 1)

test_size_days = 3
train_size = len(daily_data) - test_size_days - n_lags

X_train, X_test = X[0:train_size,:], X[train_size:len(X),:]
y_train, y_test = y[0:train_size], y[train_size:len(y)]

print(f"Data points: {len(daily_data)} | Train: {len(X_train)} | Test: {len(X_test)}")


# --- 2. Build Your Model ---
print("\n--- 2. Build Your Model ---")
model = Sequential()
model.add(LSTM(50, activation='relu', input_shape=(n_lags, 1)))
model.add(Dense(1))
model.compile(optimizer='adam', loss='mean_squared_error')
print(model.summary())


# --- 3. Train Your Model ---
print("\n--- 3. Train Your Model ---")
print("Training model (epochs=100)...")
history = model.fit(X_train, y_train, epochs=100, batch_size=1, verbose=0)
print("Training complete.")


# --- 4. Evaluate & Refine ---
print("\n--- 4. Evaluate & Refine ---")
y_pred_scaled = model.predict(X_test, verbose=0)
y_pred = scaler.inverse_transform(y_pred_scaled)
y_test_original = scaler.inverse_transform(y_test)

rmse = np.sqrt(mean_squared_error(y_test_original, y_pred))
mae = mean_absolute_error(y_test_original, y_pred)
r2 = r2_score(y_test_original, y_pred)

print(f"Test RMSE: {rmse:.2f} | MAE: {mae:.2f} | R2: {r2:.2f}")

plt.figure(figsize=(10, 6))
plt.plot(daily_data['Date'].iloc[len(daily_data) - len(y_test_original):], y_test_original, label='Actual', marker='o')
plt.plot(daily_data['Date'].iloc[len(daily_data) - len(y_pred):], y_pred, label='Predicted', marker='x')
plt.title('Actual vs. Predicted (Test Set)')
plt.xlabel('Date')
plt.ylabel('Total App Openings')
plt.xticks(rotation=45)
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.savefig('test_predictions_plot.png')
print("Generated 'test_predictions_plot.png'.")


# --- 5. Make Predictions ---
print("\n--- 5. Make Predictions ---")
last_known_data = scaled_data[-n_lags:].reshape(1, n_lags, 1)
future_predictions_scaled = []
forecast_dates = pd.date_range(start=daily_data['Date'].iloc[-1] + pd.Timedelta(days=1), periods=11)

for _ in range(11):
    next_pred_scaled = model.predict(last_known_data, verbose=0)[0, 0]
    future_predictions_scaled.append(next_pred_scaled)
    # Fix: Reshape next_pred_scaled to 3D to match last_known_data dimensions for concatenation
    last_known_data = np.append(last_known_data[:, 1:, :], [[[next_pred_scaled]]], axis=1)

future_predictions_original = scaler.inverse_transform(np.array(future_predictions_scaled).reshape(-1, 1))

print("\n--- 11-Day Forecast ---")
forecast_df = pd.DataFrame({'Date': forecast_dates, 'Predicted_App_Openings': future_predictions_original.flatten().round(2)})
print(forecast_df.to_string(index=False))

plt.figure(figsize=(12, 7))
plt.plot(daily_data['Date'], daily_data['Value'], label='Historical', marker='o', markersize=4)
plt.plot(forecast_df['Date'], forecast_df['Predicted_App_Openings'], label='Forecast', linestyle='--', marker='x', color='red')
plt.title('Historical and Forecasted App Openings')
plt.xlabel('Date')
plt.ylabel('Total App Openings')
plt.xticks(rotation=45)
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.savefig('full_forecast_plot.png')
print("Generated 'full_forecast_plot.png'.")
```
